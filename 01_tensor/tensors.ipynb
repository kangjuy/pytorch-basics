텐서의 속성

```python
tensor = torch.rand(3, 4)

print(f"shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")  # dtype = data type
print(f"Device tensor is stored on: {tensor.device}")
```

텐서 연산

```python
if torch.accelerator.is_available():  # 지금 컴퓨터에 쓸 수 있는 가속기가 있는지 확인
  tensor = tensor.to(torch.accelerator.current_accelerator()) # 텐서를 가속기(GPU)로 복사한 뒤, 그 복사본을 다시 tensor에 넣는 것
  # 이렇게 하면 이 줄 이후로 tensor를 사용하면 GPU에 있는 텐서를 사용
  # torch.accelerator.current_accelerator(): 앞에서부터 읽기! .은 "~ 안에 있는"이라는 뜻
```

```python
tensor = torch.ones(4, 4)
print(f"First row: {tensor[0]}")
print(f"First column: {tensor[:, 0]}")
print(f"Last column: {tensor[..., -1]}")  # ...: 앞의 차원들이 몇 개든 전부 ':'로 채우는 것
tensor[:, 1] = 0
print(tensor)
```

```python
t1 = torch.cat([tensor, tensor, tensor], dim = 1) # cat: concatenate(이어붙이다)
# dim = 0: 0번 차원(행) 방향, 즉 아래로 쌓임
# dim = 1: 1번 차원(열) 방향, 즉 옆으로 붙임
# n번 차원은 shape을 보기
print(t1)
```

```python
y1 = tensor @ tensor.T  # .T: 전치
y2 = tensor.matmul(tensor.T)
# @ = .matmul(): 행렬 곱 (왼쪽의 i행과 오른쪽의 j열을 한 칸씩 곱해서 전부 더한 값)
# (n, k) @ (k, m) -> (n, m)

y3 = torch.rand_like(y1)  # y1과 같은 shape([4, 4])의 텐서를 미리 만듦
torch.matmul(tensor, tensor.T, out = y3)  # 미리 만들어둔 텐서(y3)에 결과를 덮어씀

z1 = tensor * tensor
z2 = tensor.mul(tensor)
# * = mul(): 같은 위치끼리만 곱

z3 = torch.rand_like(tensor)
torch.mul(tensor, tensor, out = z3)
```

```python
agg = tensor.sum()
agg_item = agg.item() # .item: 숫자 하나만 든 텐서에서 그 값을 꺼내 파이썬 기본 숫자로 바꿈
print(agg_item, type(agg_item))
```

```python
print(f"{tensor} \n")
tensor.add_(5)  # 피연산자 자체를 변경 (여기서는 tensor)
print(tensor)

# 원래는
# a = torch.ones(2)
# b = a.add(5)

# a.add_(5) : a 자체를 바꿈
```

텐서를 NumPy 배열로 변환

```python
t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```

```python
t.add_(1)
print(f"t: {t}")
print(f"n: {n}")
# n은 건드리지 않았는데 변함
# n = t.numpy()로 만들 때 둘이 같은 메모리를 공유
# 이름만 두 개고 실제 숫자가 저장된 곳은 하나
# 공유를 끊고 싶다면 .copy()를 붙이면 됨: n = t.numpy().copy()
```

Numpy 배열을 Tensor로 변환

```python
n = np.ones(5)
t = torch.from_numpy(n) # torch 안의 from_numpy 함수에 n을 넣어라
# 오류나는 방식 - n.tensor(): n이 가진 tensor 기능을 실행해라 -> 넘파이가 먼저 생겼기 때문에 tensor를 몰라서 오류남

# 메모리 공유 O: t1 = torch.from_numpy(n) - torch안의 from_numpy 함수에 n을 넣어라
# 메모리 공유 X: t2 = torch.tensor(n)
```

```python
np.add(n, 1, out = n) # add 함수로 n에 1을 더하고, 결과를 n에 덮어씀
print(f"t: {t}")
print(f"n: {n}")
# t와 n이 메모리를 공유하고 있음
```
