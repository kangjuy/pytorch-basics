````python
import torch
import numpy as np
````

# torch.randn(32, 3, 64, 64) 만들기

````python
x = torch.randn(32, 3, 64, 64)
x.shape
````

# 채널별 평균 구하기

````python
mean = x.mean(dim = (0, 2, 3))
print(mean.shape)
````

채널별 평균을 구하려면 배치(0), 높이(2), 너비(3) 차원을 모두 평균내어 없애고 채널(1) 차원만 남겨야 하므로 dim=[0, 2, 3]을 지정

평균을 내도 그 축을 없애지 말고 크기 1로 남겨둬!
```
mean = x.mean(dim = (0, 2, 3), keepdim = True)
```

# 그 평균을 원래 텐서에서 빼기 (브로드캐스팅)

(x - mean)을 하면 에러
- 에러 이유: 브로드캐스팅은 shape을 오른쪽 끝부터 맞춰줌 -> 3이 64에 붙어서 맞지 않음

mean을 (1, 3, 1, 1) 모양으로 변경
- 방법 1: 셀 2에서 keepdim = True
- 방법 2: mean.view(1, 3, 1, 1)

````python
mean = mean.view(1, 3, 1, 1)  # 숫자 = 축의 크기
x_norm = x - mean
print(x_norm.shape)
print(x_norm.mean(dim = (0, 2, 3)))
````

# (32, 64, 64, 3)로 바꾸기 (premute)

````python
y = x.permute(0, 2, 3, 1)
print(y.shape)
````

traspose(a, b): 두 축만 바꿀 때 사용

permute(...): 축 3개 이상의 순서를 한 번에 정할 때 사용

# (32, 12288)로 펴기 (view, flatten)

```
z = y.view(32, -1)  # -1 = "나머지는 알아서 계산"
print(z.shape)
```
이렇게 하면 오류!

permute 뒤엔 에러

````python
print(y.reshape(32, -1).shape)
print(x.view(32, -1).shape)
````

view: 데이터는 그대로 두고 shape(읽는 방법)만 바꾸기

reshape: 숫자를 새 메모리에 순서대로 복사해서 새 텐서를 따로 만듦

view는 메모리 순서 그대로일 때만 되고, reshape는 필요하면 복사해서 항상 된다

# (3, 64, 64) -> (1, 3, 64, 64) (unsqueeze)

unsqueeze(n): n번 자리에 크기 1짜리 축을 하나 끼워 넣음

원소 개수는 그대로, 축만 하나 늘어남

squeeze(): 크기 1짜리 축을 뺌

괄호 안에 숫자를 넣으면 해당 번호의 축만 뺌

````python
img = torch.randn(3, 64, 64)
img4 = img.unsqueeze(0)
print(img4.shape)
````

````python
print(img.unsqueeze(1).shape)
print(img.unsqueeze(-1).shape)
print(img.squeeze(0).shape)
````

unsqueeze는 원본 img를 바꾸지 않음

위의 reshape과 동일

쌓이게 하고 싶으면 결과를 다른 변수에 담아야 함
